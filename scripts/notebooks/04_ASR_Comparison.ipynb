```python
# --- P2: SOTA (NVIDIA Canary / Parakeet) ---
print('Loading NVIDIA Canary 1B...')
try:
    import nemo.collections.asr as nemo_asr
    
    canary_model = nemo_asr.models.EncDecMultiTaskModel.from_pretrained('nvidia/canary-1b')
    canary_model.eval()
    canary_model.to('cuda')
    
    # canary_model.transcribe(['dummy_audio.wav'])
    print('Canary loaded successfully!')
except Exception as e:
    print('Lỗi tải model Canary:', e)
```
